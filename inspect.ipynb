```python
orig_path = "/cpfs01/gongshukai/weights/Ovi/Ovi/model.safetensors"
orig_path2 = "/cpfs01/gongshukai/weights/Ovi/Ovi/model_960x960.safetensors"
orig_path3 = "/cpfs01/gongshukai/weights/Ovi/Ovi/model_960x960_10s.safetensors"
save_path = "/cpfs01/gongshukai/step_distillation/logs/ovi_original.pt"
save_path2 = "/cpfs01/gongshukai/step_distillation/logs/ovi_original_960x960.pt"
save_path3 = "/cpfs01/gongshukai/step_distillation/logs/ovi_original_960x960_10s.pt"

import safetensors.torch
import torch

state_dict = safetensors.torch.load_file(orig_path, device='cpu')
state_dict2 = safetensors.torch.load_file(orig_path2, device='cpu')
state_dict3 = safetensors.torch.load_file(orig_path3, device='cpu')

def remap_ovi_state_dict_for_refactored(state_dict):
    new_state_dict = {}
    for k, v in state_dict.items():
        new_key = k
        # 规则 1: video_model.blocks.X.* -> fusion_blocks.X.vid_block.*
        if k.startswith("video_model.blocks."):
            parts = k.split('.')
            block_idx = parts[2]
            remaining_path = ".".join(parts[3:])
            new_key = f"fusion_blocks.{block_idx}.vid_block.{remaining_path}"
        # 规则 2: audio_model.blocks.X.* -> fusion_blocks.X.audio_block.*
        elif k.startswith("audio_model.blocks."):
            parts = k.split('.')
            block_idx = parts[2]
            remaining_path = ".".join(parts[3:])
            new_key = f"fusion_blocks.{block_idx}.audio_block.{remaining_path}"
        # 规则 3: video_model.* -> video_*
        elif k.startswith("video_model."):
            new_key = k.replace("video_model.", "video_")
        # 规则 4: audio_model.* -> audio_*
        elif k.startswith("audio_model."):
            new_key = k.replace("audio_model.", "audio_")
        
        new_state_dict[new_key] = v
    return new_state_dict

state_dict_refactored = remap_ovi_state_dict_for_refactored(state_dict)
state_dict_refactored2 = remap_ovi_state_dict_for_refactored(state_dict2)
state_dict_refactored3 = remap_ovi_state_dict_for_refactored(state_dict3)

torch.save(state_dict_refactored, save_path)
print(f"Saved refactored state dict to {save_path}")
torch.save(state_dict_refactored2, save_path2)
print(f"Saved refactored state dict to {save_path2}")
torch.save(state_dict_refactored3, save_path3)
print(f"Saved refactored state dict to {save_path3}")
```

```python
from transformers import AutoTokenizer, T5Tokenizer

# 替换为你服务器上 umt5-xxl 的实际路径
# 例如: "/cpfs01/gongshukai/weights/Ovi/Wan2.2-TI2V-5B/google/umt5-xxl"
# 如果本地没有，可以填 "google/umt5-xxl" 让它从 HuggingFace 下载 (需要联网)
MODEL_PATH = "/cpfs01/gongshukai/weights/Ovi/Wan2.2-TI2V-5B/google/umt5-xxl" 

def check_length(prompt):
    try:
        # 加载 Tokenizer (T5 系列通常用 T5Tokenizer 或 AutoTokenizer)
        tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
    except Exception as e:
        print(f"加载 Tokenizer 失败: {e}")
        return

    # 编码
    # Wan2.2 代码中设置了 seq_len=512，这里我们要看原始长度，所以不截断
    tokens = tokenizer(prompt, add_special_tokens=True)["input_ids"]
    
    length = len(tokens)
    print("=" * 50)
    print(f"Prompt: {prompt[:50]}...")
    print(f"Token Count: {length}")
    
    if length > 512:
        print(f"❌ 警告: 超出 512 限制！超出了 {length - 512} 个 Token。")
        print("建议: 精简 Prompt 或移除不重要的修饰词。")
    else:
        print(f"✅ 正常: 占用 {length}/512，剩余 {512 - length} 空间。")
    print("=" * 50)

if __name__ == "__main__":
    # 把你的长 Prompt 粘贴在这里
    my_prompt = "The video opens with a medium close-up shot of a woman with dark hair, visible from her chest up. She is positioned on the right side of the frame, looking down and slightly to the left, listening intently. Her lips are slightly parted, suggesting she is about to speak. Behind her, a second woman with lighter, longer hair is partially visible and out of focus, standing further back and to the left. The background woman says, <S>It connects people from all across the world.<E> The foreground woman then responds, <S>Oh. So, it's like a fancy telephone?<E> She smiles towards the end of the clip. The background consists of a wall with a textured, possibly yellow or beige wallpaper. A white curtain hangs on the left side of the frame, letting in soft light. A dark wooden shelf with decorative items is visible on the wall behind the women. The camera remains static throughout the video, focusing on the foreground woman's subtle facial expressions. <AUDCAP>A clear, articulate female voice speaks the first sentence, slightly from the background. A second clear, articulate female voice responds with a questioning, then slightly amused and joyful tone. There is no background music or prominent sound effects, only natural room ambient sound.<ENDAUDCAP>"
    
    check_length(my_prompt)
```
